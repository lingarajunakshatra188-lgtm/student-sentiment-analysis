```python
# Sector names
sectors = [
    'Placement Training',
    'Canteen Quality',
    'Campus Environment',
    'Teaching Quality',
    'Technical Event Participation',
    'Hostel Facilities',
    'Internet Facilities',
    'Lab Facilities'
]

# Satisfaction scores
Avg_Satisfaction_score = [4.90, 5.12, 5.19, 5.35, 5.39, 5.53, 5.58, 5.70]

# Create scatter plot
plt.figure(figsize=(12,6))
plt.scatter(sectors, Avg_Satisfaction_score)

# Add titles
plt.title("Sector vs Average Satisfaction")
plt.xlabel("Sectors")
plt.ylabel("Average Satisfaction")

# Rotate sector names
plt.xticks(rotation=20)
# Show plot
plt.show()
```
